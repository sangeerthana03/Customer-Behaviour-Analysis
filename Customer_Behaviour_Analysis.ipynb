```python
df['Review Rating']=df.groupby('Category') ['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
label=['Young Adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=label)
```

```python
# Discounted vs Non-Discounted Purchases by Category

category_discount = (
    df.groupby(['category', 'discount_applied'])['customer_id']
      .count()
      .reset_index(name='purchase_count')
)

print(category_discount)

# Chart
plt.figure(figsize=(9, 5))

bar = sns.barplot(
    data=category_discount,
    x='category',
    y='purchase_count',
    hue='discount_applied'
)

plt.bar_label(bar.containers[0], fmt='%d')
plt.bar_label(bar.containers[1], fmt='%d')

plt.title('Discounted vs Non-Discounted Purchases by Category')
plt.xlabel('Category')
plt.ylabel('Number of Purchases')
plt.show()
```

```python
#"How many high-rated purchases does each category receive from subscribers vs non-subscribers?"

high_rated = df[df['review_rating'] >= 4.0]

plt.figure(figsize=(10, 5))

bar = sns.countplot(
    data=high_rated,
    x='category',
    hue='subscription_status',
    color='teal'
)

for container in bar.containers:
    plt.bar_label(container)

plt.title('High-Rated Purchases by Category and Subscription Status')
plt.xlabel('Category')
plt.ylabel('Number of High-Rated Purchases')
plt.legend(title='Subscription Status')

plt.show()
```

```python
# Revenue from Repeat Customers by Category

repeat_category_revenue = (
    df[df['previous_purchases'] > 5]
    .groupby('category')['purchase_amount']
    .sum()
    .sort_values(ascending=False)
)

print(repeat_category_revenue)

# Chart
plt.figure(figsize=(8, 5))

bar = sns.barplot(
    x=repeat_category_revenue.index,
    y=repeat_category_revenue.values
)

plt.bar_label(bar.containers[0], fmt='${:,.0f}')

plt.title('Revenue from Repeat Customers by Category')
plt.xlabel('Category')
plt.ylabel('Total Revenue (USD)')
plt.show()
```

```python
plt.figure(figsize=(8, 4))

corr = df[
    ['age', 'purchase_amount', 'review_rating',
     'previous_purchases', 'purchase_frequency_days']
].corr()

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f'
)

plt.title('Correlation Between Customer Behaviour Variables')
plt.show()
```
